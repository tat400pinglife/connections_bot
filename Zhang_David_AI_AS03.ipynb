```python
# node
class Node:
    def __init__(self, name):
        self.name = name
        self.neighbors = {} 

    def add_neighbor(self, neighbor, cost):
        self.neighbors[neighbor] = cost

    def return_neighbors(self):
        return self.neighbors

# map
class Map:
    def __init__(self):
        self.nodes = {} 

    def add_node(self, name):
        if name not in self.nodes:
            self.nodes[name] = Node(name)
            print(f"Added node '{name}'")
        else:
            print(f"Node '{name}' already exists")

    def add_edge(self, from_node, to_node, dist):
        if from_node in self.nodes and to_node in self.nodes:
            # assuming that you can go back and forth, which im pretty sure seems pretty logical
            self.nodes[from_node].add_neighbor(to_node, dist)
            self.nodes[to_node].add_neighbor(from_node, dist)
            print(f"Edge mapped: {from_node} to {to_node} (Distance: {dist})")
        else:
            print("atleast one of the nodes is missing")

    def print_map(self):
        for name, node_obj in self.nodes.items():
            print(f"{name} is connected to: {node_obj.return_neighbors()}")

```

```python
import pandas as pd
# make map nodes
my_map = Map()

node_mapping = pd.read_excel('RomaniaMap.xlsx')
# print(df.head())
# print(df["City"])

for city in node_mapping['City']:
    my_map.add_node(city)
    
```

```python
import pandas as pd

def load_sld_matrix(filepath, sheet_name="SLD MI"):
    try:
        df = pd.read_excel(filepath, sheet_name=sheet_name, index_col = 0)
        
        sld_dict = df.to_dict(orient='index')
        
        print(f"loaded SLD matrix for {len(sld_dict)} nodes.")
        return sld_dict

    except Exception as e:
        print(f"Error loading SLD Matrix: {e}")

sld_matrix = load_sld_matrix('RomaniaMap.xlsx', sheet_name='SLD MI')

print(sld_matrix)
```

```python
import heapq

def a_star(graph, start, goal, sld_matrix):

    if start not in graph.nodes or goal not in graph.nodes:
        print(f"Error: Start '{start}' or goal '{goal}' is missing from the map!")
        return None

    open_list = []
    closed_set = set()
    g_scores = {name: float('inf') for name in graph.nodes}
    g_scores[start] = 0
    came_from = {}
    
    start_h = sld_matrix[start][goal]
    heapq.heappush(open_list, (start_h, start))

    print(f"\nStarting A* Search: {start} to {goal} ...")

    while open_list:
        # Pop the node with the lowest f_score
        current_f, current_name = heapq.heappop(open_list)
        
        if current_name in closed_set:
            continue
        print(f"\nCurrently at {current_name} (Current f-score: {current_f})")

        closed_set.add(current_name)
        # Check if we reached the finish line
        if current_name == goal:
            print(f"reached goal, returning path")
            path = []
            while current_name in came_from:
                path.append(current_name)
                current_name = came_from[current_name]
            path.append(start)
            return path[::-1]

        current_node = graph.nodes[current_name]
        neighbors = current_node.return_neighbors()
        
        if not neighbors:
            print(" Ran out of edge connections")

        # Explore all neighbors
        for neighbor_name, travel_distance in neighbors.items():
            print(f"Evaluating path to: {neighbor_name}")
            
            tentative_g = g_scores[current_name] + travel_distance
            print(f" Tentative g score is {tentative_g}")
            # If this is a faster route, process it
            if tentative_g < g_scores[neighbor_name]:
                print(f" g score is better(less than) other neighbor(s)")
                # add city to path
                came_from[neighbor_name] = current_name
                g_scores[neighbor_name] = tentative_g
                

                h_score = sld_matrix[neighbor_name][goal]
                f_score = tentative_g + h_score
                print(f" h score is {h_score}, f score is {f_score}")

                heapq.heappush(open_list, (f_score, neighbor_name))
                
                print(f" Path to {neighbor_name} | g: {tentative_g}, h: {h_score}, f: {f_score}")
                print(f" Added {neighbor_name} to the open list.")
            else:
                print(f" Known path to {neighbor_name} is already cheaper or equal.")

    print("\n PATH DNE")
    return None
```

```python
print(a_star(my_map, "Oradea", "Urziceni", sld_matrix))
```

# Conclusion

For this specific A* search from Oradea to Urziceni obtained path ['Oradea', 'Sibiu', 'Rimnicu Vilcea', 'Pitesti', 'Bucharest', 'Urziceni']. 

The search algorithm performs by evaluating each path/edge and calculating the potential f-score via the g and h-score distances from the data. Each path is pushed into heap queue (open_list) such that the item with highest priority is the path with the lowest f-score. It then takes the greedy approach and goes to the next location with the lowest f-score, this then repeats until it reaches the target.
