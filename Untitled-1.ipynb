```python
import pandas as pd

df = pd.read_csv('https://archive.ics.uci.edu/ml/machine-learning-databases/iris/iris.data', header = None)
```

```python
# •	A) sepal length B) sepal width C) petal length D) petal width
df.columns = ['sepal length', 'sepal width', 'petal length', 'petal width', 'species']

print(df.head())
```

```python
from sklearn.linear_model import SGDClassifier
import matplotlib.pyplot as plt
from sklearn.model_selection import cross_val_score
from sklearn.preprocessing import LabelEncoder
from matplotlib.colors import ListedColormap
import numpy as np

# one feature (petal width)
X = df.iloc[:, 3:4] 
# target species variable
y = df.iloc[:, -1]


loss_functions = ['log_loss', 'hinge', 'modified_huber', 'squared_hinge', 'perceptron']
accuracies = []

y_text = df['species']   
# ---------------------------------------------------

# 3. Convert text labels ('Iris-setosa', etc.) to numbers (0, 1, 2)
encoder = LabelEncoder()
y = encoder.fit_transform(y_text)

# 5. Train and collect the CV scores
for loss in loss_functions:
    model = SGDClassifier(loss=loss, max_iter=2000, random_state=42)
    scores = cross_val_score(model, X, y, cv=5)
    accuracies.append(scores.mean() * 100)

fig, axes = plt.subplots(1, 3, figsize=(15, 5), sharey=True)

# Define custom colors: Light colors for the background, bold for the dots
cmap_light = ListedColormap(['#DDA0DD', '#20B2AA', '#F0E68C']) # Purple, Teal, Yellow (Light)
cmap_bold = ListedColormap(['#800080', '#008080', '#DAA520'])  # Purple, Teal, Yellow (Dark)

# Create a "fake" grid of data points covering the entire X-axis
x_min, x_max = X.iloc[:, 0].min() - 0.5, X.iloc[:, 0].max() + 0.5
y_min, y_max = -0.5, 2.5
xx, yy = np.meshgrid(np.linspace(x_min, x_max, 500), np.linspace(y_min, y_max, 50))

# 5. Train each model and plot its specific "thoughts"
for ax, loss in zip(axes, loss_functions):
    
    # Initialize and train the model on all data
    model = SGDClassifier(loss=loss, max_iter=2000, random_state=42)
    model.fit(X, y)
    
    # Ask the model to predict every single point on our "fake" grid
    Z = model.predict(xx.ravel().reshape(-1, 1))
    Z = Z.reshape(xx.shape)
    
    # Color in the background based on the model's predictions
    ax.contourf(xx, yy, Z, cmap=cmap_light, alpha=0.4)
    
    # Plot the true, actual data points on top
    ax.scatter(X, y, c=y, cmap=cmap_bold, edgecolor='k', s=60)
    
    # Formatting
    ax.set_title(f'Decision Boundary\n({loss})')
    ax.set_xlabel('Feature Value')

# Set up the Y-axis labels on the very first chart
axes[0].set_ylabel('True Class')
axes[0].set_yticks([0, 1, 2])
axes[0].set_yticklabels(encoder.classes_) 

plt.tight_layout()
plt.show()
```

```python
from sklearn.linear_model import SGDClassifier
from sklearn.metrics import accuracy_score

# two features (petal width and petal length)
X = df.iloc[:, 2:4] 
# target species variable
y = df.iloc[:, -1]


loss_functions = ['log_loss', 'hinge', 'modified_huber', 'squared_hinge', 'perceptron']

print("Training...\n")

for loss in loss_functions:
    model = SGDClassifier(loss=loss, max_iter=2000, random_state=42)

    model.fit(X, y)
    y_pred = model.predict(X)
    accuracy = accuracy_score(y, y_pred)
    
    print(f"Loss Function: {loss:<15} | Accuracy: {accuracy * 100:.2f}%")

print("cross-validation...\n")
from sklearn.model_selection import cross_val_score
for loss in loss_functions:
    model = SGDClassifier(loss=loss, max_iter=2000, random_state=42)
    scores = cross_val_score(model, X, y, cv=5)
    mean_accuracy = scores.mean()
    
    print(f"Loss Function: {loss:<15} | True Accuracy: {mean_accuracy * 100:.2f}%")
```
